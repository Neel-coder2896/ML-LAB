## Practical 1: Student Marks Analyzer
### Problem Definition
Store student data as a list of tuples. Remove duplicate marks, calculate average marks
and store result in a dictionary.
### Sample Input
students = [
(1, "Amit", [78, 85, 90, 85]),
(2, "Neha", [88, 92, 79, 88]),
(3, "Ravi", [70, 75, 80, 75])
]
### Sample Output
{1: ('Amit', 84.33), 2: ('Neha', 86.33), 3: ('Ravi', 75.0)}

```python
students=[
(1, "Amit", [78, 85, 90, 85]),
(2, "Neha", [88, 92, 79, 88]),
(3, "Ravi", [70, 75, 80, 75])
]

result={}
for studentid,name,marks in students:
    nonduplicate=set(marks)

    averagemarks=sum(nonduplicate)/len(nonduplicate)

    result[studentid]=(name,round(averagemarks,2))

print(result)
```

## Practical 3: Email Domain Extractor
### Problem Definition
Extract email domains and count occurrences.
### Sample Input
emails = ['a@gmail.com', 'b@yahoo.com', 'c@gmail.com']
### Sample Output
{'gmail.com': 2, 'yahoo.com': 1}

```python
emails = ['a@gmail.com', 'b@yahoo.com', 'c@gmail.com']
count_domain={}
for email in emails:
    result=email.split('@')[1] # [a, gmail.com]
    if result in count_domain:
        count_domain[result]+=1
    else:
        count_domain[result]=1
print(count_domain)        
            


```

## Practical 4: Shopping Cart Billing
### Problem Definition
Calculate item-wise cost and total bill.
### Sample Input
cart = [('apple',2), ('banana',3)]
prices = {'apple':50, 'banana':10}
### Sample Output
{'apple': 100, 'banana': 30}
Total Bill: 130


```python
cart = [('apple',2), ('banana',3)]
prices = {'apple':50, 'banana':10}
result={}

for name,piece in cart:
    for price in prices:
        result={name,price*piece}
print(result)        


```

## Practical 5: Character Position Index
### Problem Definition
Store positions of each character in a string.
### Sample Input
s = "banana"
### Sample Output
In [14]:
{'b': [0], 'a': [1, 3, 5], 'n': [2, 4]}

```python
s="banana"
ch_postions={}

for index,char in enumerate(s):
    if char in ch_postions:
        ch_postions[char].append(index)
    else:
        ch_postions[char]=[index]    
print(ch_postions)        
        

```
