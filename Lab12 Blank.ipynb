# Step 1: Import Libraries

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

from sklearn.datasets import make_moons, make_blobs
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import NearestNeighbors
```

# Step 2: Generate dummy data using make_blobs

```python
x, y = make_blobs(n_samples=300,centers=3)
```

# Step 3: Plot Generated data 

```python
plt.scatter(x[::,0],x[::,1])
plt.show()
```

# Step 4: Apply Hierarchical Clustering

```python
from sklearn.cluster import AgglomerativeClustering
```

```python
model=AgglomerativeClustering(n_clusters=4)
```

```python
model.fit(x)
```

```python
model.labels_
```

```python
plt.scatter(x[::,0],x[::,1],c=model.labels_)
plt.show()
```

# Step 5: Display Dendrogram

```python
from scipy.cluster.hierarchy import dendrogram,linkage
```

```python
linked=linkage(x,method='ward')
dendrogram(linked)
plt.show()
```

# Step 6: Generate dummy data using make_moons

```python
x,y=make_moons(n_samples=300,noise=0.20,random_state=50)
x=StandardScaler().fit_transform(x)
```

# Step 7: Plot Generated data

```python
plt.scatter(x[:,0],x[:,1],)
plt.show()
```

# Step 8: apply DBSCAN Clustering 

```python
from sklearn.cluster import DBSCAN
```

```python
dbscan=DBSCAN(eps=0.2,min_samples=5)
dbscan.fit(x)
```

```python
DBSCAN(eps=0.2)
```

```python
dbscan.labels_
```

```python
plt.scatter(x[:,0],x[:,1],c=dbscan.labels_)
plt.show()
```

# Step 9: apply K Mean Clustering on same data

```python
from sklearn.cluster import KMeans

kmeans=KMeans(n_clusters=3,random_state=42)
kmeans_label=kmeans.fit_predict(x)
centerdata=kmeans.cluster_centers_

plt.scatter(x[:,0],x[:,1],c=kmeans_label)
plt.scatter(centerdata[:,0],centerdata[:,1],marker="x")
plt.show()
```
